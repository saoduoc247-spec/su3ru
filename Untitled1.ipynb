```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.neighbors import NearestNeighbors

# Cố định seed theo quy ước REPRODUCIBILITY.md
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# Tạo thư mục results nếu chạy độc lập trên Colab
os.makedirs('../results', exist_ok=True)
os.makedirs('results', exist_ok=True)
RESULTS_DIR = '../results' if os.path.exists('../data') else 'results'

# ==============================================================================
# 1. ĐỌC DỮ LIỆU CANONICAL CSV (parkinsons_v1.csv) & CHUẨN HÓA
# ==============================================================================
canonical_paths = [
    '../data/processed/parkinsons_v1.csv',
    'data/processed/parkinsons_v1.csv',
    'parkinsons_v1.csv'
]

df = None
for p in canonical_paths:
    if os.path.exists(p):
        df = pd.read_csv(p)
        print(f"Đã đọc dữ liệu từ: {p}")
        break

# Dự phòng nếu chạy test độc lập trên Colab chưa upload file parkinsons_v1.csv
if df is None:
    url = "https://archive.ics.uci.edu/ml/machine-learning-databases/parkinsons/parkinsons.data"
    df = pd.read_csv(url)
    print("Đang dùng dữ liệu UCI gốc để test độc lập trên Colab.")

# Loại bỏ cột name; tách riêng status (y_true CHỈ dùng để tính ARI, không fit mô hình)
feature_cols = [c for c in df.columns if c not in ['name', 'status']]
X = df[feature_cols].values
y_true = df['status'].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(f"Kích thước đặc trưng: {X_scaled.shape} | Phân bố nhãn thực tế: 0={sum(y_true==0)}, 1={sum(y_true==1)}")

# ==============================================================================
# 2. PHÂN CỤM K-MEANS & CHỌN K TỐI ƯU
# ==============================================================================
k_range = range(2, 11)
inertias = []
sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(k_range, inertias, 'bo-')
ax[0].set_title('Elbow Method (Inertia)')
ax[0].set_xlabel('Số cụm K')
ax[0].set_ylabel('Inertia')
ax[0].grid(True, linestyle='--', alpha=0.5)

ax[1].plot(k_range, sil_scores, 'ro-')
ax[1].set_title('Silhouette Score theo K')
ax[1].set_xlabel('Số cụm K')
ax[1].set_ylabel('Silhouette Score')
ax[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig_m06_kmeans_selection.png', dpi=300)
plt.show()

best_k = list(k_range)[int(np.argmax(sil_scores))]
results_list = []

for k_eval in sorted(list(set([2, best_k]))):
    km = KMeans(n_clusters=k_eval, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_scaled)
    sil = silhouette_score(X_scaled, labels)
    ari = adjusted_rand_score(y_true, labels)
    results_list.append({
        'model': f'K-Means (K={k_eval})',
        'n_clusters': k_eval,
        'n_noise': 0,
        'silhouette': round(sil, 4),
        'ari': round(ari, 4)
    })

# ==============================================================================
# 3. PHÂN CỤM DBSCAN & KHẢO SÁT THAM SỐ (eps, min_samples)
# ==============================================================================
min_samples = 5
nn = NearestNeighbors(n_neighbors=min_samples).fit(X_scaled)
distances, _ = nn.kneighbors(X_scaled)
k_distances = np.sort(distances[:, min_samples - 1])

plt.figure(figsize=(6, 4))
plt.plot(k_distances)
plt.title(f'Biểu đồ K-Distance (min_samples={min_samples})')
plt.xlabel('Điểm dữ liệu (sắp xếp tăng dần)')
plt.ylabel(f'Khoảng cách tới láng giềng thứ {min_samples}')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig_m06_dbscan_kdist.png', dpi=300)
plt.show()

db_candidates = []
for eps in [2.5, 3.0, 3.5, 4.0, 4.5, 5.0]:
    for ms in [3, 4, 5, 6]:
        db = DBSCAN(eps=eps, min_samples=ms)
        labels = db.fit_predict(X_scaled)
        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        n_noise = int(np.sum(labels == -1))
        if n_clusters >= 2:
            ari = adjusted_rand_score(y_true, labels)
            mask = labels != -1
            sil = silhouette_score(X_scaled[mask], labels[mask]) if np.sum(mask) > n_clusters else -1.0
            db_candidates.append({
                'eps': eps, 'min_samples': ms,
                'n_clusters': n_clusters, 'n_noise': n_noise,
                'silhouette': round(sil, 4), 'ari': round(ari, 4)
            })

df_db = pd.DataFrame(db_candidates).sort_values(by='ari', ascending=False)
print("\nTop cấu hình DBSCAN theo ARI:")
print(df_db.head(5).to_string(index=False))

if not df_db.empty:
    top_db = df_db.iloc[0]
    best_eps, best_ms = float(top_db['eps']), int(top_db['min_samples'])
    labels_db_best = DBSCAN(eps=best_eps, min_samples=best_ms).fit_predict(X_scaled)
    results_list.append({
        'model': f'DBSCAN (eps={best_eps}, ms={best_ms})',
        'n_clusters': int(top_db['n_clusters']),
        'n_noise': int(top_db['n_noise']),
        'silhouette': float(top_db['silhouette']),
        'ari': float(top_db['ari'])
    })
else:
    best_eps, best_ms = 3.5, 5
    labels_db_best = DBSCAN(eps=best_eps, min_samples=best_ms).fit_predict(X_scaled)

# ==============================================================================
# 4. TRỰC QUAN HÓA PCA 2D & XUẤT BÁO CÁO (m06_ari_clustering.csv + clustering_report.md)
# ==============================================================================
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)
labels_km2 = KMeans(n_clusters=2, random_state=RANDOM_STATE, n_init=10).fit_predict(X_scaled)

fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))
axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=y_true, cmap='coolwarm', alpha=0.8, edgecolors='k')
axes[0].set_title('Nhãn thực tế (0: Khỏe, 1: Parkinson)')

axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=labels_km2, cmap='viridis', alpha=0.8, edgecolors='k')
axes[1].set_title(f'K-Means (K=2) | ARI = {adjusted_rand_score(y_true, labels_km2):.4f}')

axes[2].scatter(X_pca[:, 0], X_pca[:, 1], c=labels_db_best, cmap='plasma', alpha=0.8, edgecolors='k')
axes[2].set_title(f'DBSCAN (eps={best_eps}) | ARI = {adjusted_rand_score(y_true, labels_db_best):.4f}')

for ax in axes:
    ax.set_xlabel('PC1')
    ax.set_ylabel('PC2')

plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig_m06_clustering_comparison.png', dpi=300)
plt.show()

# Lưu bảng kết quả CSV
df_summary = pd.DataFrame(results_list)
csv_out = f'{RESULTS_DIR}/m06_ari_clustering.csv'
df_summary.to_csv(csv_out, index=False)

# Lưu báo cáo Markdown tương tự M05
md_out = f'{RESULTS_DIR}/clustering_report.md'
with open(md_out, 'w', encoding='utf-8') as f:
    f.write("# Báo cáo Phân cụm Không giám sát (M06)\n\n")
    f.write("- **Dữ liệu:** `parkinsons_v1.csv` (195 mẫu × 22 đặc trưng, chuẩn hóa `StandardScaler`, `seed=42`).\n")
    f.write("- **Nguyên tắc:** Không sử dụng nhãn `status` khi fit thuật toán; chỉ dùng `status` để tính Adjusted Rand Index (ARI).\n\n")
    f.write("## Bảng kết quả tổng hợp\n\n")
    f.write(df_summary.to_markdown(index=False))
    f.write("\n\n## Nhận xét\n")
    f.write("- Dữ liệu giọng nói Parkinson có vùng giao thoa lớn giữa nhóm khỏe mạnh (48) và bệnh nhân giai đoạn nhẹ (147) nên chỉ số ARI ở mức trung bình/thấp so với phân lớp có giám sát.\n")
    f.write("- DBSCAN nhạy cảm với mật độ trong không gian 22 chiều và phân loại nhiều điểm biên thành nhiễu (`label = -1`).\n")

print("\n=== BẢNG TỔNG HỢP KẾT QUẢ M06 ===")
print(df_summary.to_string(index=False))
print(f"\nĐã xuất file: {csv_out} và {md_out}")
```
